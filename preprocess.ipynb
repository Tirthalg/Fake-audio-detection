```python
import os
import pandas as pd

fake_audio_dir = './for-norm/training/fake'
original_audio_dir = './for-norm/training/real'
data = []

def process_audio_files(directory, label):
    for file_name in os.listdir(directory):
        file_path = os.path.join(directory, file_name)
        data.append([file_path, label])

process_audio_files(fake_audio_dir, 'fake')
process_audio_files(original_audio_dir, 'original')
```

```python

fake_audio_dir = './for-norm/testing/fake'
original_audio_dir = './for-norm/testing/real'


process_audio_files(fake_audio_dir, 'fake')
process_audio_files(original_audio_dir, 'original')
```

```python

fake_audio_dir = './for-norm/validation/fake'
original_audio_dir = './for-norm/validation/real'

process_audio_files(fake_audio_dir, 'fake')
process_audio_files(original_audio_dir, 'original')
```

```python

df = pd.DataFrame(data, columns=['file_path', 'label'])
output_csv_path = 'audio_data_labels.csv'
df.to_csv(output_csv_path, index=False)

print(f"CSV file created at: {output_csv_path}")
```
